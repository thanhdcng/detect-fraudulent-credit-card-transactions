```python
import pandas as pd

# Read the CSV file
df = pd.read_csv('fraud.csv')

# Display the first few rows of the dataframe
df.head()
```

```python
# Check the data types of columns
df.dtypes
```

```python
# Display basic information about the dataset
df.info()
```

```python
# Display summary statistics
df.describe()
```

```python
# Check for missing values 
missing_values_column = df.isnull().sum()
print("Missing values per column:")
print(missing_values_column)

# Calculate the percentage of missing values per column
missing_percentage = (missing_values_column / len(df)) * 100
print("Percentage of missing values per column:")
print(missing_percentage)
```

```python
# Check for duplicate rows
duplicates = df.duplicated()
print(duplicates.sum())
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
```

```python
# Count the occurrences of each value in the 'is_fraud' column
count_fraud = df['is_fraud'].value_counts(sort=True)

# Plot the counts as a bar chart
ax = count_fraud.plot(kind = 'bar', rot = 0)
plt.title('Fraud vs Non-Fraud Transactions')
plt.xlabel('Fraudulent')
plt.ylabel('Count')

# Add the count values on top of the bars
for p in ax.patches:
    ax.annotate(str(p.get_height()), (p.get_x() * 1.005, p.get_height() * 1.005))

# Show the plot
plt.show()
```

```python
# Plot boxplot for 'amt' column
plt.boxplot(df['amt'])
plt.title('Boxplot of amt Column')
plt.xlabel('amt')
plt.ylabel('Values')
plt.show()
```

```python
# Display the 'amt' column values greater than 10000
amt_values = df[df['amt'] > 10000]['amt']
print(amt_values)
```
